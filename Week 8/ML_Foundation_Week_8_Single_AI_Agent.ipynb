````python
# Import Required Libraries

import re
import math
````

# 🧮 Calculator Tool

The Calculator Tool evaluates mathematical expressions entered by the user.

Examples:

- calculate 15 + 10
- calculate 20 / 5
- calculate (5 + 3) * 2

The tool returns the calculated result or an appropriate error message.

````python
# Calculator Tool

def calculator(expression):

    try:

        result = eval(expression)

        return str(result)

    except Exception as e:

        return f"Calculation Error: {e}"
````

# 🔑 Keyword Extraction Tool

The Keyword Extraction Tool identifies important words from a sentence by:

- converting text to lowercase
- removing punctuation
- removing short words
- returning meaningful keywords

This demonstrates basic Natural Language Processing (NLP).

````python
# Keyword Extraction Tool

def extract_keywords(text):

    text = re.sub(r"[^\w\s]", "", text)

    words = text.lower().split()

    keywords = [word for word in words if len(word) > 3]

    return keywords
````

# 🤖 AI Agent

The AI Agent acts as the decision-making component of the system.

It analyzes the user's query and routes it to the appropriate tool.

### Supported Tasks

✔ Calculator

✔ Keyword Extraction

✔ General Conversation

The response is returned as a structured JSON object.

## ⚙️ Agent Implementation
👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
def agent(query):

    query = query.strip()

    query_lower = query.lower()

    try:

        # Calculator Route
        if query_lower.startswith("calculate"):

            expression = query[len("calculate"):].strip()

            result = calculator(expression)

            return {

                "type": "calculation",

                "query": query,

                "result": result

            }

        # Keyword Extraction Route
        elif "keyword" in query_lower:

            text = query

            if "from" in query_lower:

                text = query.split("from", 1)[1].strip()

            result = extract_keywords(text)

            return {

                "type": "keywords",

                "query": query,

                "result": result

            }

        # General Conversation
        else:

            return {

                "type": "general",

                "query": query,

                "result": "I can help with calculations and keyword extraction."

            }

    except Exception as e:

        return {

            "type": "error",

            "query": query,

            "result": str(e)

        }


# -----------------------
# Test Cases
# -----------------------

print(agent("Calculate 15 + 10"))

print()

print(agent("Extract keywords from Artificial Intelligence is transforming healthcare."))

print()

print(agent("Hello Agent"))


# -----------------------
# Interactive Mode
# -----------------------

print("\nAI Agent Started (type 'exit' to quit)\n")

while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":

        print("Agent: Goodbye!")

        break

    print(agent(user_input))

    print()
````
